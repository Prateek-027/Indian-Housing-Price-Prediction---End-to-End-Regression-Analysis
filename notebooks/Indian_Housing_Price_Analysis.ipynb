# 4. Data Preprocessing

```python
# Checking the invalid/impossible values as:
# area_sqft<=0
# bedrooms<=0
# bathrooms<=0
# floor < 0
# total_floors <=0
# distance <0
# amenities score outside expected range
# crime_index outside expected range
```

```python
print('Area_sqft <= 0:', (df['area_sqft'] <= 0).sum())
print('Bedrooms <= 0: ', (df['bedrooms'] <= 0).sum())
print('Bathrooms <= 0: ', (df['bathrooms'] <= 0).sum())
print('Floor < 0: ', (df['floor'] < 0).sum())
print('Total Floors <= 0: ', (df['total_floors'] <= 0).sum())
print('Floor>Total_floors: ', (df['floor'] > df['total_floors']).sum())
print('Metro Distance < 0: ', (df['distance_to_metro_km'] < 0).sum())
print('School Distance < 0: ', (df['distance_to_school_km'] < 0).sum())
print('Hospital Distance < 0: ', (df['distance_to_hospital_km'] < 0).sum())
print('Amenities Score Outside Range: ', ((df['amenities_score'] < 0) | (df['amenities_score'] > 10)).sum())
print('Crime Index Outside Range: ', ((df['crime_index'] < 0) | (df['crime_index'] > 100)).sum())
```

```python
x=df.drop(columns = ['property_id', 'price_lakh'])
y = df['price_lakh']
print('X Shape: ', x.shape)
print('Y Shape: ', y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.20, random_state = 42)
print('X Train Shape: ', x_train.shape)
print('X Test Shape: ', x_test.shape)
print('Y Train Shape: ', y_train.shape)
print('Y Test Shape: ', y_test.shape)
```

```python
numerical_features = ['area_sqft', 'bedrooms', 'bathrooms', 'floor', 'total_floors', 'property_age_years', 'parking_spaces', 'distance_to_metro_km', 'distance_to_school_km','distance_to_hospital_km', 'amenities_score', 'crime_index','monthly_maintenance_inr']
categorical_features = ['city', 'locality_tier', 'property_type', 'furnishing', 'ownership_type']
```

```python
print('Numerical Features: ', len(numerical_features))
print('Categorical Features: ', len(categorical_features))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
# Imputation decision:
# Numerical missing values: Median (because several variables show skewness or higher extreme values or both)
# Categorical missing values: Mode
```

```python
numerical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy = 'median')),
    ('scaler', StandardScaler())
])
```

```python
categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy = 'most_frequent')),
    ('encoder', OneHotEncoder(drop = 'first', handle_unknown = 'ignore'))
])
```

```python
preprocessor = ColumnTransformer([
    ('num', numerical_pipeline, numerical_features),
    ('cat', categorical_pipeline, categorical_features)
])
```

# 6. Cross-validation and Hyperparameter Tuning

```python
# Cross Val Score - TO check whether the performance is reasonably consistent across different training/validation splits.
```

```python
from sklearn.model_selection import cross_val_score
```

```python
cv_scores = cross_val_score(
    gb_model,
    x_train,
    y_train,
    cv = 5,
    scoring = 'r2'
)
```

```python
print('R2 Score: ', cv_scores)
print('Mean CV R2: ', round(cv_scores.mean(), 3))
print('Std CV R2: ', round(cv_scores.std(), 3))
```

```python
# Hyperparameter Tuning
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'model__n_estimators': [100, 200],
    'model__learning_rate': [0.05, 0.1],
    'model__max_depth': [2,3]
}
```

```python
grid_search = GridSearchCV(
    gb_model,
    param_grid,
    cv=5,
    scoring='r2',
    n_jobs=1
)
```

```python
grid_search.fit(x_train, y_train)
```

```python
print('Best Parameters: ', grid_search.best_params_)
print('Best CV R2: ', round(grid_search.best_score_, 3))
```

```python
best_model = grid_search.best_estimator_
best_pred = best_model.predict(x_test)
```

```python
print('Final MAE: ', round(mean_absolute_error(y_test, best_pred), 2))
print('Final RMSE: ', round(np.sqrt(mean_squared_error(y_test, best_pred)), 2))
print('Final R2: ', round(r2_score(y_test, best_pred), 2))
```

# 8. Feature Importance

```python
# Get the fitted preprocessing step
preprocessor_fitted = best_model.named_steps['preprocessor']

# Get feature names after preprocessing
feature_names = preprocessor_fitted.get_feature_names_out()

# Get feature importance from Gradient Boosting
importance = best_model.named_steps['model'].feature_importances_

# Creating a DataFrame
feature_importance = pd.DataFrame({
    'feature': feature_names,
    'importance': importance
})

# Sort by importance
feature_importance = feature_importance.sort_values(
    'importance',
    ascending = False
)

feature_importance.head(15)
```

```python
plt.figure(figsize = (8,6))
plt.barh(
    feature_importance.head(10)['feature'][::-1],
    feature_importance.head(10)['importance'][::-1]
)
plt.xlabel('Feature Importance')
plt.ylabel('Feature')
plt.title('Top 10 Feature Importances')
```

```python
# Feature importance indicates that area_sqft is the most influential feature in predicting property price, 
# followed by location-related features such as Mumbai and Prime locality.
# This is consistent with the EDA, where property size and location showed strong relationships with price.
# Feature importance represents the contribution of features to model predictions and should not be interpreted as causation.
```

```python
models = {
    'Baseline' : baseline_pred,
    'Linear Regression' : linear_pred,
    'Ridge' : ridge_pred,
    'Lasso' : lasso_pred,
    'Random Forest' : rf_pred,
    'Gradient Boosting' : gb_pred,
    'Tuned Gradient Boosting' : best_pred
}
```

```python
results = []
for model_name, predictions in models.items():
    mae=mean_absolute_error(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    r2= r2_score(y_test, predictions)
    results.append({
        'Model': model_name,
        'MAE' : round(mae,2),
        'RMSE' : round(rmse, 2),
        'R2' : round(r2,3)
    })
comparison_df = pd.DataFrame(results)
comparison_df
```
